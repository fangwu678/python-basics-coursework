# Functions

This notebook shows a simple function with a return value.

It is based on one of my original Python classroom exercises.

```python
# Define a function to calculate the average of a list
def average(myList):
    total = sum(myList)  # Use sum() to get the total
    average = total / len(myList)  # Use len() to get the number of items
    return average

# Use the function
scores = [7, 23, 56, 89]
averageScore = average(scores)
print("Average score:", averageScore)
```
